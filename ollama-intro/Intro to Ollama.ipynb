# Intro to Ollama via Langchain

## Imports

```python
from langchain.agents import create_agent
from dotenv import load_dotenv
import os
import ipywidgets as widgets
from IPython.display import display
```

```python
# Load from Environment
load_dotenv('.env', override=True)
LLM_HOST = os.getenv('OLLAMA_HOST')
LLM_MODEL = os.getenv('OLLAMA_MODEL')
```

```python
agent = create_agent(
    model=f"ollama:{LLM_MODEL}",
    system_prompt="You are a helpful model , which attempts to answer user's queries. In case of doubts or ambiguity, ask follow-ups don't assume and answer"
)
```

```python
def invoke_model(text: str) -> None:
    model_query = {
        "messages": [{
            "role": "user",
            "content": text
        }]
    }
    result = agent.invoke(model_query)
    return result
```

```python
invoke_model("Hello")
```

```python
# Create the text box widget
text_box = widgets.Text(
    value='',
    placeholder='Type your query here...',
    description='Input:',
    disabled=False,
    continuous_update=False
)
# Output widget to display the result cleanly
out = widgets.Output()
submit_button = widgets.Button(
    description="Submit", button_style="success"
)
```

```python
def invoke_model_and_show_output(text: str) -> None:
    result = invoke_model(text)
    with out:
        out.clear_output()
        print(f"{result["messages"][-1].content_blocks}")
```

```python
submit_button.on_click(lambda x: invoke_model_and_show_output(text_box.value))
display(text_box, submit_button, out)
```
